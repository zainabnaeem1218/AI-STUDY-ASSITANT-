```python
score = 0

answer1 = input("Q1: What is AI? ")

if "computer" in answer1.lower() or "intelligence" in answer1.lower():
    print("Correct! ✅")
    score += 1
else:
    print("Good try!")

answer2 = input("Q2: What programming language are we using? ")

if "python" in answer2.lower():
    print("Correct! ✅")
    score += 1
else:
    print("Good try!")

answer3 = input("Q3: What does CS stand for? ")

if "computer science" in answer3.lower():
    print("Correct! ✅")
    score += 1
else:
    print("Good try!")

print("\nYour final score is:", score, "/ 3")

if score == 3:
    print("Excellent work! 🌟")
elif score == 2:
    print("Good job! Keep practising! 👍")
else:
    print("Keep learning and try again! 📚")
```
